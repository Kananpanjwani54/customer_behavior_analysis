```python
bins = [0, 18, 25, 35, 45, 55, 65, 100]

labels = [
    "Under 18",
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65",
    "65+"
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)

df["Age Group"].value_counts().sort_index()
```

```python
df.groupby("Age Group", observed=True)["Purchase Amount (USD)"].agg(
    ["count", "mean", "sum"]
)
```

```python
Q1 = df["Purchase Amount (USD)"].quantile(0.25)
Q3 = df["Purchase Amount (USD)"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Purchase Amount (USD)"] < lower) |
    (df["Purchase Amount (USD)"] > upper)
]

print("Number of outliers:", len(outliers))
```
